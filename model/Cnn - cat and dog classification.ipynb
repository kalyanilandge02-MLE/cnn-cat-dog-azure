### Data Preprocessing

```python
image_size = (224,224)

data_generator = ImageDataGenerator(rescale = 1/255, validation_split = 0.2)

train_generator = data_generator.flow_from_directory(r"C:\Users\kalya\Downloads\data-20251229T120206Z-3-001\data",target_size = image_size,
                                                     batch_size = 32, class_mode= "binary",
                                                     subset = "training")

val_generator = data_generator.flow_from_directory(r"C:\Users\kalya\Downloads\data-20251229T120206Z-3-001\data", target_size = image_size,
                                                   batch_size = 32, class_mode = 'binary',
                                                   subset = "validation")
```

### Model Building

```python
from tensorflow.keras.models import Sequential

model = Sequential()
model.add(Conv2D(filters = 32,
                 kernel_size = (3,3),
                 activation = 'relu',
                 input_shape = image_size + (3,)))
model.add(MaxPool2D(pool_size = (2,2)))

#Flatten Layer
model.add(Flatten())

# model.add(Dense(units = 4, activation = 'relu'))

# Dense Layers
model.add(Dense(units = 64, activation = 'relu'))

# Output Layers
model.add(Dense(units = 1 , activation = 'sigmoid'))
model.summary()

early_stopping = EarlyStopping(monitor = 'accuracy',patience = 5, verbose = 1)
model.compile(optimizer = 'adam',loss = 'binary_crossentropy',metrics = ['accuracy'])
history = model.fit(train_generator, validation_data = val_generator, callbacks = early_stopping, epochs = 50, verbose = 2)
```

### Data Augmentation

```python
Without Data Augmentation:
  1st Epoch (11/11) >> Same images
  2nd Epoch >> Same images
  3rd Epoch >> Same images

With Data Augmentation:
  1st Epoch >> Different images
  2nd Epoch >> Different images
  3rd Epoch >> Different images

To reduce overfitting

Data Augmentation:(Train Data)-On the fly
  Rotation(skewness)
  Shift(right and left)
  Zoom in/out
  Flip(right/left)
```

```python
image_size = (224,224)

data_generator = ImageDataGenerator(rescale = 1/255, validation_split = 0.2,
                                    width_shift_range = 0.1,
                                    height_shift_range = 0.1,
                                    zoom_range = 0.15,
                                    horizontal_flip = True,
                                    vertical_flip = False,
                                    shear_range = 0.1 
                                   )
train_generator = data_generator.flow_from_directory(r"C:\Users\kalya\Downloads\data-20251229T120206Z-3-001\data",target_size = image_size,
                                                     batch_size = 32, class_mode = "binary",
                                                     subset = "training")

val_generator = data_generator.flow_from_directory(r"C:\Users\kalya\Downloads\data-20251229T120206Z-3-001\data",target_size = image_size,
                                                   batch_size = 32,class_mode = "binary",
                                                   subset = "validation")

```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.callbacks import EarlyStopping

model = Sequential()
# Convolution Layers
model.add(Conv2D(filters = 32,
                 kernel_size=(3,3),
                 activation='relu',
                 input_shape = image_size + (3,) ))

model.add(MaxPool2D(pool_size=(2, 2)))

#Flatten Layer
model.add(Flatten())

# Dense Layer
model.add(Dense(units = 64, activation = 'relu'))

# Output Layers
model.add(Dense(units = 1, activation = 'sigmoid'))
model.summary()

early_stopping = EarlyStopping(monitor='accuracy', patience=5, verbose = 1)
model.compile(optimizer='adam', loss = 'binary_crossentropy', metrics = ['accuracy'])
history = model.fit(train_generator, validation_data=val_generator,
callbacks = early_stopping, epochs = 50, verbose = 2)
```

### Hyperparameter Tuning

```python
def build_model(hp):
    model = Sequential()

    # Convolution Layer1
    activation_1 = hp.Choice("kernel_initializer_1",['relu','tanh'])
    kernel_initializer_1 = hp.Choice("kernel_initializer_21",['glorot_uniform','he_normal'])
    kernel_size_1 = hp.Choice("kernel_size_1",[3,5])

    model.add(Conv2D(filters = 32,
                     kernel_size = kernel_size_1,
                     activation = activation_1,
                     kernel_initializer = kernel_initializer_1,
                     input_shape = image_size + (3,)))

    dropout_rate_1 = hp.Float("droupout_rate_1",min_value = 0.0, max_value = 0.51, step = 0.1)
    model.add(Dropout(rate = dropout_rate_1))

    model.add(MaxPool2D(pool_size = (2,2)))
    model.add(BatchNormalization(beta_initializer = 'zeros',gamma_initializer = 'ones'))

    # Flatten Layer
    model.add(Flatten())

    # Dense Layer
    units_2 = hp.Choice("units_2",[8,16,32,64,128])
    activation_2 = hp.Choice("activation_2",['relu','tanh'])
    kernel_initializer_2 = hp.Choice("kernel_initializer_2",['glorot_uniform','he_normal'])

    model.add(Dense(units = units_2, activation = activation_2,
                    kernel_initializer = kernel_initializer_2))
    model.add(BatchNormalization(beta_initializer = 'zeros',gamma_initializer = 'ones'))

    dropout_rate_2 = hp.Float("dropout_rate_2",min_value = 0.0, max_value = 0.51 , step = 0.1)
    model.add(Dropout(rate = dropout_rate_2))

    # output Layer
    model.add(Dense(units = 1, activation = 'sigmoid'))

    # select Optimizer
    optimizers_choice = hp.Choice("optimizer",['adam','sgd','rmsprop'])
    learning_rate = hp.Float("learning_rate",min_value = 0.001,max_value = 0.1, step = 0.005)
    if optimizers_choice == 'adam':
        optmizer = keras.optimizers.Adam(learning_rate = learning_rate)
    elif optimizers_choice == 'sgd':
        optmizer = keras.optimizers.SGD(learning_rate = learning_rate)
    elif optimizers_choice == 'rmsprop':
        optmizer = keras.optimizers.RMSprop(learning_rate = learning_rate)

    model.compile(optimizer = optmizer , loss = 'binary_crossentropy',metrics = ['accuracy'])
    return model
tuner = kt.RandomSearch(build_model,
                        objective = "accuracy",
                        max_trials = 10,
                        project_name = "CNN_Binary_Classification")

tuner.search(train_generator, validation_data = val_generator,
             epochs = 10,verbose = 2)

                     
    
```
